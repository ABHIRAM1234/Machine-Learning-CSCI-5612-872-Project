```python
import math
import random

# point1 = [3, 4]
# point2 = [0, 0]
# i = 0:
# → (3 - 0) ** 2 = 9
# → distance = 0 + 9 = 9

# i = 1:
# → (4 - 0) ** 2 = 16
# → distance = 9 + 16 = 25

# Final step:
# → math.sqrt(25) = 5

# Euclidean distance calculation between two points
def euclidean_distance(point1, point2):
    distance = 0
    for i in range(len(point1)):
        distance += (point1[i] - point2[i]) ** 2
    return math.sqrt(distance)



# K-Means algorithm implementation
def kmeans(data, k, max_iterations=100):
    centroids = random.sample(data, k)  # Randomly initialize k centroids using the points in the data. instead of generating centroids randomly from space (which might land outside the data), it ensures each initial centroid is an actual data point 
    
    # K-Means clustering loop
    for _ in range(max_iterations): 
        clusters = [[] for _ in range(k)]  # Create empty clusters -> clusters = [[], [], [], ....]
        
        # Assign data points to the nearest centroid
        for point in data: # data = [[2,3], [8,7], [1, 4], ...]. We need to loop over each point in the data and calculate euclidean distance with all centroids
            distances = [euclidean_distance(point, centroid) for centroid in centroids] #  distance of (2,3) from centroid k1, k2, k3....
            nearest_centroid = distances.index(min(distances)) # calculates the index number of the centroid which is the closest to the point. nearest_centroid = distances.index(1.0)  # → 0 So, this point [2,3] belongs to cluster 0.
            clusters[nearest_centroid].append(point) # You add the current point to the corresponding cluster.
            
        
        prev_centroids = centroids.copy()  # Store previous centroids
        
        # Update centroids as the mean of each cluster
        for i in range(k):
            if clusters[i]:
                centroids[i] = [sum(feature) / len(clusters[i]) for feature in zip(*clusters[i])]
        
        # Check for convergence
        if prev_centroids == centroids:
            break
    
    return clusters, centroids
```
